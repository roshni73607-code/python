```python
#minimum 8character
#the alphabet must be between[a-z]
#atleast one alphabet should be uppercase
#at least one digit
#at least on special symbol from
#write program to check password is valid or not
a=input("enter a password: ")
#
u=0
l=0
sp=0
d=0
total=0
if len(a)>=8:
    for i in a:
  
        if i.isupper():
            u+=1
        if i.islower():
            l+=1
        if i.isdigit():
             d+=1
        if (i=="_" or i=="@" or i=="$") :
            sp+=1

            total+=u+l+d+sp
    if u!=0 and l!=0 and d!=0 and sp!=0 and (len(a)==total):
        print("password is valid")
    else:
        print("not valid")
                   
else:
    print("not valid")
        
```

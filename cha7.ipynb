# modules
- A python module is A file containing pyhthon defination and statement .a module can define function ,class, and variable .is makes the code logically organized.

```python
%%writefile mymodule.py
def add(a,b):
    print(a+b)
def sub(a,b):
    print(a-b)
def mul(a,b):
    print(a*b)
def div(a,b):
    print(a/b)
def power(a,b):
    print(a**b)
def floor(a,b):
    print(a//b)
def modulo(a,b):
    print(a%b)
```

```python
import mymodule
mymodule.add(2,4)
dir(mymodule)
```

```python
import mymodule as m
m.sub(8,9)
```

```python
from mymodule import add,mul
add(2,4)
mul(8,9)
sub(7,9)
```

```python
from mymodule import *
add(2,4)
mul(8,9)
sub(7,9)
```

```python
from mymodule import add as addition,mul as multi
addition(2,4)
mul(8,9)
sub(7,9)
```

## date-represent a celender date (year,month,day) 
## time-represent a time of day(hour,minute,second,msec)
## datetime-combines both date and time in formation

```python
from datetime import date
today=date.today()
year,month,date=today.year,today.month,today.day
print(today)
print(year,month,date)
```

```python
from datetime import date
d=date(2026,11,30)
print(d)
```

```python
from datetime import date
today=date.today()
print(today.strftime("%d %b %Y (%A)"))
print(today.strftime("%d %b %Y (%a)"))
print(today.strftime("%d %b %y (%A)"))
print(today.strftime("%d %B %Y (%a)"))
```

```python
from datetime import date
today=date(2023,11,22)
print(today.strftime("%d/%m/%Y"))
print(today.strftime("%d-%m-%y"))
```

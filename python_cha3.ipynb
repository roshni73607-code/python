```python
#variable length argument
def sum(*n):#* argument
    total=0#n is tuple
    for n1 in n:
        total+=n1
    print("total:",total)
sum(10)
sum(10,20,30)
sum(10,40,50,68)
```

```python
def f1(n1,*s):#* argument
    print(n1)
  
    for s1 in s:
       #first is n1 other values are s
        print(s1)
f1(10)
f1(10,20,30)
f1(10,40,50,68)
```

```python
def f1(*s,n1):#* argument
   
  
    for s1 in s:#all values are s
        print(s1)
    print(n1)
f1(10,20,n1=40)
f1(10,20,30)
f1(10,"A","B","C")
```

```python
def display(**n):#keyword argumment
    print(n)#not using n
    for k,v in n.items():
        
        print(k,"=",v)
display(m1=100,m2=200,m3=300)
display(rho=1,name="100",marks=200,subject="python")
      
```
